
# Heart Disease Analysis Using Custom Dataset

This notebook performs exploratory data analysis and classification modeling using a heart disease dataset. The goal is to understand the features that contribute to heart disease and build a predictive model.


```python
#Data Preprocessing
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import numpy as np
# Load the dataset
df = pd.read_csv('heart-new.csv') 

# Check for missing values
df.isnull().sum()

# Impute missing numerical values with mean
numerical_cols = df.select_dtypes(include=[np.number]).columns
imputer = SimpleImputer(strategy='mean')
df[numerical_cols] = imputer.fit_transform(df[numerical_cols])

# Encode binary categorical variables
binary_cols = ['Smoking', 'AlcoholDrinking', 'Stroke', 'DiffWalking', 'Diabetic', 
               'PhysicalActivity', 'Asthma', 'KidneyDisease', 'SkinCancer']
for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

# One-hot encode multi-class categorical variables
df = pd.get_dummies(df, columns=['Sex', 'AgeCategory', 'Race', 'GenHealth'], drop_first=True)
print(df.columns.tolist())
# Feature Scaling
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df.drop('HeartDisease', axis=1))
X = pd.DataFrame(scaled_features, columns=df.drop('HeartDisease', axis=1).columns)
y = df['HeartDisease'].map({'Yes': 1, 'No': 0})
```

```python
# 📊 Plot Heart Disease Count
import matplotlib.pyplot as plt
import seaborn as sns
label_col = 'HeartDisease' if 'HeartDisease' in df.columns else 'target'

plt.figure(figsize=(6, 4))
sns.countplot(x=label_col, data=df)
plt.title('Heart Disease Count')
plt.xlabel(label_col)
plt.ylabel('Count')
plt.show()
```

```python
# Simplify diabetic values into categories
df['Diabetic'] = df['Diabetic'].replace({
    'No': 0,
    'No, borderline diabetes': 1,
    'Yes (during pregnancy)': 1,
    'Yes': 2
})
```

```python
# Check for missing values
print(df.isnull().sum())

# Impute missing numerical values with mean
from sklearn.impute import SimpleImputer
import numpy as np

numerical_cols = df.select_dtypes(include=[np.number]).columns
imputer = SimpleImputer(strategy='mean')
df[numerical_cols] = imputer.fit_transform(df[numerical_cols])
```

```python
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

# Impute missing values with most frequent value (works for mixed categorical/numerical)
imputer = SimpleImputer(strategy='most_frequent')
X_imputed = imputer.fit_transform(X)

# Now split
X_train, X_test, y_train, y_test = train_test_split(X_imputed, y, test_size=0.2, random_state=42, stratify=y)
```

```python
# Logistic Regression Model with balanced class 
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt

# Logistic Regression Model with balanced class weights
log_reg = LogisticRegression(max_iter=1000, class_weight='balanced')
log_reg.fit(X_train, y_train)

# Prediction and Evaluation
y_pred = log_reg.predict(X_test)
y_proba = log_reg.predict_proba(X_test)[:, 1]

# Output results
print("Logistic Regression (Balanced):")
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("Classification Report:")
print(classification_report(y_test, y_pred))
print(f"ROC AUC Score: {roc_auc_score(y_test, y_proba):.4f}")

# Calculate Precision, Recall, and F1 Score
precision_log_reg = precision_score(y_test, y_pred)
recall_log_reg = recall_score(y_test, y_pred)
f1_log_reg = f1_score(y_test, y_pred)

# Plot the metrics
plt.figure(figsize=(6, 4))
plt.bar(['Precision', 'Recall', 'F1 Score'], [precision_log_reg, recall_log_reg, f1_log_reg], color=['skyblue', 'salmon', 'lightgreen'])
plt.title('Logistic Regression (Balanced) - Precision, Recall, F1 Score')
plt.ylim(0, 1)
plt.grid(axis='y')
plt.show()

# Optionally, plot the confusion matrix using matplotlib
cm_log_reg = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
plt.imshow(cm_log_reg, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix for Logistic Regression')
plt.colorbar()
tick_marks = range(2)
plt.xticks(tick_marks, ['0', '1'])
plt.yticks(tick_marks, ['0', '1'])

thresh = cm_log_reg.max() / 2.
for i in range(cm_log_reg.shape[0]):
    for j in range(cm_log_reg.shape[1]):
        plt.text(j, i, format(cm_log_reg[i, j], 'd'),
                 horizontalalignment="center",
                 color="white" if cm_log_reg[i, j] > thresh else "black")

plt.ylabel('True label')
plt.xlabel('Predicted label')
plt.tight_layout()
plt.show()


```

```python
# Decision Trees
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_score, recall_score

# Decision Tree Model with class_weight to handle imbalance
dt_model = DecisionTreeClassifier(random_state=42, class_weight='balanced')
dt_model.fit(X_train, y_train)

# Prediction and Evaluation
y_pred_dt = dt_model.predict(X_test)
y_proba_dt = dt_model.predict_proba(X_test)[:, 1]

print("Decision Tree (Balanced):")
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))
print("Classification Report:")
print(classification_report(y_test, y_pred_dt))
print(f"ROC AUC Score: {roc_auc_score(y_test, y_proba_dt):.4f}")

# Extract Precision, Recall, and F1-Score for each class
report = classification_report(y_test, y_pred_dt, output_dict=True)

# Extract precision, recall, f1-score values
precision_dt = [report['0']['precision'], report['1']['precision']]
recall_dt = [report['0']['recall'], report['1']['recall']]

# Plot Precision and Recall for each class
labels = ['Class 0', 'Class 1']

x = range(len(labels))

plt.figure(figsize=(10,6))

plt.bar(x, precision_dt, width=0.4, label='Precision', align='center')
plt.bar(x, recall_dt, width=0.4, label='Recall', align='edge')

plt.xlabel('Classes')
plt.ylabel('Score')
plt.title('Precision and Recall for Decision Tree (Balanced)')
plt.xticks(x, labels)
plt.legend()

plt.show()

# Optionally, plot the confusion matrix using matplotlib
cm = confusion_matrix(y_test, y_pred_dt)

plt.figure(figsize=(8, 6))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix for Decision Tree (Balanced)')
plt.colorbar()
tick_marks = range(2)
plt.xticks(tick_marks, ['0', '1'])
plt.yticks(tick_marks, ['0', '1'])

thresh = cm.max() / 2.
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, format(cm[i, j], 'd'),
                 horizontalalignment="center",
                 color="white" if cm[i, j] > thresh else "black")

plt.ylabel('True label')
plt.xlabel('Predicted label')
plt.tight_layout()
plt.show()
```

```python
#Random Forest
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Random Forest with class_weight to handle imbalance
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced'  # This adjusts weights inversely to class frequencies
)
rf_model.fit(X_train, y_train)

# Prediction and Evaluation
y_pred_rf = rf_model.predict(X_test)
y_proba_rf = rf_model.predict_proba(X_test)[:, 1]

print("Random Forest (with class_weight='balanced'):")
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
print("Classification Report:")
print(classification_report(y_test, y_pred_rf))
print(f"ROC AUC Score: {roc_auc_score(y_test, y_proba_rf):.4f}")

# Extract Precision, Recall, and F1-Score for each class
report_rf = classification_report(y_test, y_pred_rf, output_dict=True)

# Extract precision, recall, f1-score values
precision_rf = [report_rf['0']['precision'], report_rf['1']['precision']]
recall_rf = [report_rf['0']['recall'], report_rf['1']['recall']]

# Plot Precision and Recall for each class
labels = ['Class 0', 'Class 1']

x = range(len(labels))

plt.figure(figsize=(10,6))

plt.bar(x, precision_rf, width=0.4, label='Precision', align='center')
plt.bar(x, recall_rf, width=0.4, label='Recall', align='edge')

plt.xlabel('Classes')
plt.ylabel('Score')
plt.title('Precision and Recall for Random Forest Classifier')
plt.xticks(x, labels)
plt.legend()

plt.show()

# Optionally, plot the confusion matrix using matplotlib
cm_rf = confusion_matrix(y_test, y_pred_rf)

plt.figure(figsize=(8, 6))
plt.imshow(cm_rf, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix for Random Forest Classifier')
plt.colorbar()
tick_marks = range(2)
plt.xticks(tick_marks, ['0', '1'])
plt.yticks(tick_marks, ['0', '1'])

thresh = cm_rf.max() / 2.
for i in range(cm_rf.shape[0]):
    for j in range(cm_rf.shape[1]):
        plt.text(j, i, format(cm_rf[i, j], 'd'),
                 horizontalalignment="center",
                 color="white" if cm_rf[i, j] > thresh else "black")

plt.ylabel('True label')
plt.xlabel('Predicted label')
plt.tight_layout()
plt.show()

```
